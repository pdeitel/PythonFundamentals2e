# 9.3.1 Writing to a Text File: Introducing the with Statement CheckPoint

1. **(IPython Session)** Create a `grades.txt` file and write to it the following three records consisting of student IDs, last names and letter grades:
```
1 Diaz A
2 Laine B
3 Ali A
```

#### Answer:

````python
with open('grades.txt', mode='x', encoding='utf-8') as grades:
    grades.write('1 Diaz A\n')
    grades.write('2 Laine B\n')
    grades.write('3 Ali A\n')
````
