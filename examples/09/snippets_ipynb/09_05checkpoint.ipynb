# 9.5 Objects-Natural Case Study: JSON Serialization and Deserialization Checkpoint

1. **(IPython Session)** Create a JSON file named `grades.json` and write into it the following dictionary:
```python
grades_dict = {
    'gradebook': [
        {'student_id': 1, 'name': 'Diaz', 'grade': 'A'},
        {'student_id': 2, 'name': 'Laine', 'grade': 'B'},
        {'student_id': 3, 'name': 'Ali', 'grade': 'A'}
    ]
}
```
Then, read the file and display its pretty-printed JSON.

#### Answer

````python
import json
````

````python
grades_dict = {
    'gradebook':[
        {'student_id': 1, 'name': 'Red', 'grade': 'A'},
        {'student_id': 2, 'name': 'Green', 'grade': 'B'},
        {'student_id': 3, 'name': 'White', 'grade': 'A'}
    ]
}
````

````python
with open('grades.json', mode='x', encoding='utf-8') as grades:
    json.dump(grades_dict, grades, ensure_ascii=False)
````

````python
!cat grades.json
````

````python
with open('grades.json', mode='r', encoding='utf-8') as grades:
    print(json.dumps(json.load(grades), indent=4))
````
