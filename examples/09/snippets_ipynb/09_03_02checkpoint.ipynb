# 9.3.2 Reading Data from a Text File Checkpoint

1. **(IPython Session)** Read the file `grades.txt` that you created in the previous section's Checkpoint and display it in columns with the column heads `'ID'`, `'Name'` and `'Grade'`. Left-align the `'ID'` column in a field of four characters and the `'Name'` column in a field of seven characters.

#### Answer:

```python
with open('grades.txt', mode='r', encoding='utf-8') as grades:
    print(f'{"ID":<4}{"Name":<7}{"Grade"}')
    for record in grades:
        student_id, name, grade = record.split()
        print(f'{student_id:<4}{name:<7}{grade}')
```
