# 10.10 Operator Overloading Checkpoint

**(IPython Session)** Modify class `Complex` to support the minus (`-`) operator, then test the `-` and `-=` operators.

#### Answer: 

The new method definition (located in `complexnumber2.py`) is:
```python
def __sub__(self, right: Complex) -> Complex:
    """Overloads the - operator."""
    return Complex(self.real - right.real,
                   self.imaginary - right.imaginary)
```

````python
from complexnumber2 import Complex
````

````python
x = Complex(real=2, imaginary=4)
````

````python
y = Complex(real=5, imaginary=-1)
````

````python
x - y
````

````python
x -= y # Python evaluates this as x = x - y
````
