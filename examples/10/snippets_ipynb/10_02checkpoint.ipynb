# 10.2 Custom Class `Account` Checkpoint

**(IPython Session)** Add a `withdraw` method to class `Account`. If the withdrawal `amount` is greater than the `balance`, raise a `ValueError` indicating that the withdrawal `amount` must be less than or equal to the `balance`. If the withdrawal `amount` is less than or equal to `0.00`, raise a `ValueError` indicating that the withdrawal `amount` must be positive. If the withdrawal `amount` is valid, subtract it from the `balance` attribute. Next, in an IPython session, create an `Account` object, then test its `withdraw` method with
- a valid withdrawal `amount`,
- an invalid withdrawal `amount` greater than the `balance` and
- an invalid negative withdrawal `amount`.

#### Answer: 

We duplicated `account.py` and named it `account_enhanced.py` for this Checkpoint’s solution. The new `Account` method is:
```python
def withdraw(self, amount: Decimal) -> None:
    """Withdraw money from the account."""

    # raise an exception for an invalid withdrawal amount
    if amount > self.balance:
        raise ValueError('amount must be <= balance.')
    elif amount <= Decimal('0.00'):
        raise ValueError('amount must be positive.')

    self.balance -= amount
```

This method must be indented in the `Account` class definition at the same level as the `__init__` and `deposit` methods. The following IPython session tests `withdraw`:

````python
from account_enhanced import Account
````

````python
from decimal import Decimal
````

````python
account1 = Account('Sofia Lopez', Decimal('50.00'))
````

````python
account1.withdraw(Decimal('20.00'))
````

````python
account1.balance
````

````python
account1.withdraw(Decimal('100.00'))
````

````python
account1.withdraw(Decimal('-10.00'))
````
