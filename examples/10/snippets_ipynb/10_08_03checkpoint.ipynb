**(What Does This Code Do?)** Explain what the following statement from class `Employee`’s `earnings` method does:
```python
return self.compensation_model.earnings()
```

Answer: This statement gets the `CompensationModel` object referenced by the `Employee`’s `compensation_model` property, calls that object’s `earnings` method and returns the result. Python determines at execution time which `CompensationModel` subclass’s `earnings` method to call, based on the type of the object the property references.
